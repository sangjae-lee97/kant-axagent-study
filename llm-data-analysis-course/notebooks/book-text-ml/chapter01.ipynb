```python
products = [
    {"name":"키보드" , "price":45000 , "stock":12, "category":"전자기기"},
    {"name":"마우스" , "price":45000 , "stock":0, "category":"전자기기"},
    {"name":"모니터" , "price":45000 , "stock":4, "category":"전자기기"},
    {"name":"책상" , "price":45000 , "stock":0, "category":"가구"},
    {"name":"의자" , "price":45000 , "stock":7, "category":"가구"},
    {"name":"키보드" , "price":45000 , "stock":5, "category":"전자기기"},
    {"name":"마우스" , "price":45000 , "stock":3, "category":"전자기기"},
    {"name":"의자" , "price":45000 , "stock":3, "category":"가구"},
]
products
```

```python
for_sales = []

found = False

for product in products:
    if product["stock"] >= 1:
        found = False
        for sale in for_sales:
            if sale["name"] == product["name"]:
                sale["stock"] += product["stock"]
                found = True
                break

        if not found:
            for_sales.append(product.copy())

for_sales
```

### 실습 6. 판매가를 숫자로 바꾸기

 

Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.

 

16,200

17,820

7,200

 

사람이 보기에는 숫자처럼 보이지만 pandas에서는 문자열로 읽힐 수 있습니다.

 

문자열 상태에서는 평균, 합계 등의 숫자 계산을 제대로 할 수 없습니다.

 

따라서 쉼표를 제거하고 숫자 자료형으로 변환합니다.

```python
# 1. 판매가에서 쉼표 제거
df_books["판매가"] = df_books["판매가"].str.replace(",", "")

# 2. 숫자형으로 변환
df_books["판매가"] = df_books["판매가"].astype(int)

# 3. 변환 후 자료형 확인
print(df_books["판매가"].dtype)
```

### 실행 결과 요약

- 문자열 전처리: `판매가` 컬럼에 포함된 쉼표를 제거했다.
- 자료형 변환: 문자열 형태의 `판매가`를 정수형으로 변환했다.
- 변환 확인: 자료형이 `int64`로 정상적으로 변경되었다.
- 분석 가능: 이제 `판매가`를 이용해 평균, 합계 등의 숫자 계산을 할 수 있다.

### 실습 7. 발행일을 날짜형으로 바꾸기

 

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

 

20260701

20260831

20020120

 

이 값은 연도-월-일 정보를 가지고 있지만 바로 날짜 계산에 사용하기 어려울 수 있습니다.

 

날짜 자료형으로 변환하면 연도, 월, 날짜 등을 쉽게 분석할 수 있습니다.

```python
# 발행일을 날짜 자료형으로 변환
df_books["발행일"] = pd.to_datetime(
    df_books["발행일"],
    format="%Y%m%d"
)

# 변환 후 자료형 확인
print(df_books["발행일"].dtype)

# 앞의 5개 값 확인
print(df_books["발행일"].head())
```

### 실행 결과 요약

- 날짜 전처리: `발행일` 컬럼을 `YYYYMMDD` 형식에서 날짜 자료형으로 변환했다.
- 자료형 변환: `발행일`의 dtype이 `datetime64[us]`로 변경되었다.
- 값 확인: `20140519` 같은 값이 `2014-05-19`처럼 읽기 쉬운 날짜 형식으로 변환되었다.
- 변환 확인: 앞의 5개 값을 확인한 결과 날짜 변환이 정상적으로 적용되었다.

### 실습 8. 문자열 앞뒤 공백 정리하기

 

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.

 

문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

"소설"

이번에 정리할 컬럼

저자

출판사

분야


```python
# 앞뒤 공백을 제거할 컬럼 목록
text_columns = ["상품명", "저자", "출판사", "분야"]

# 각 컬럼의 앞뒤 공백 제거
for col in text_columns:
    df_books[col] = df_books[col].str.strip()

# 처리 후 앞의 5개 값 확인
for col in text_columns:
    print(f"\n{col}")
    print(df_books[col].head())
```
