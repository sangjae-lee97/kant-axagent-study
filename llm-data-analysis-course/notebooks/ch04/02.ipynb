```python
from course_utils.paths import get_project_root, get_data_dir
import pandas as pd

print("프로젝트 루트 :", get_project_root())
print("데이터 폴더 :", get_data_dir())

customers = pd.read_csv(get_data_dir() / "raw/customers.csv")
orders = pd.read_csv(get_data_dir() / "raw/orders.csv")

print(customers.info())
print(orders.info())
```

```python
# products
# orders_items

products = pd.read_csv(get_data_dir() / "raw/products.csv")
order_items = pd.read_csv(get_data_dir() / "raw/order_items.csv")

print(products.info())
print(order_items.info())
```

```python
# orders의 상위 5개 row
print(orders.head())
```

```python
# 주문 상태의 종류는 몇가지 이고 각 상태별 주문 수량
ret = orders["order_status"].value_counts(dropna=False)
print(type(ret))
print(ret)
# 결제 수단의 종류는 몇가지 이고 각 상태별 주문 수량
pet = orders["payment_method"].value_counts(dropna=False)
print(type(pet))
print(pet)
```

```python
# 도시별 회원수
print(customers["city"].value_counts(dropna=False))
```

```python
print(customers["city"].unique)
```

```python
# 서울, 부산, 인천 도시의 고객들은 누구인가?
# customer_city = customers[(customers["city"] == "인천") | (customers["city"] == "부산") | (customers["city"] == "서울") ]
# print(customer_city)

city_customers = customers[customers["city"].isin(["서울", "부산", "인천"])]
print(city_customers.head())
```

```python
print(city_customers["city"].value_counts())
```

```python
# 컬럼추가(파생 컬럼)
# 라인 합계 금액 : line_total = quantity * unit_price
# print(order_items.info())
order_items["line_total"] = (order_items["quantity"] * order_items["unit_price"])
print(order_items)
```

```python
print(orders["order_id"].duplicated().sum())
```

```python
# print(order_items.head(10))
# order_status를 함께 출력 하려면 -> merge
order_sales = order_items.merge(
    orders[["order_id", "customer_id", "order_date", "order_status"]],
    on = "order_id",
    how = "left",
    validate = "many_to_one"
)
print(order_sales)
```
