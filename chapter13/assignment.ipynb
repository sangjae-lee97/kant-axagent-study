```python
sentence = "파이썬 기초 공부 파이썬 문법 연습 파이썬 코드 작성 파이썬 실습 진행 파이썬 에러 해결 파이썬 복습 완료 파이썬 프로젝트 시작 데이터 수집 작업 데이터 정제 과정 데이터 처리 연습 데이터 분석 진행 데이터 시각화 완료 데이터 저장 관리 코딩 기초 훈련 코딩 문법 이해 코딩 실습 반복 코딩 테스트 준비 코딩 알고리즘 풀이 공부 계획 수립 공부 시간 확보 공부 집중 유지 공부 내용 정리 공부 습관 형성 성공 목표 설정 성공 의지 확립 성공 습관 유지 성공 결과 달성 파이썬 데이터 코딩 공부 성공 목표"
words = sentence.split()
word_dict = {}

for word in words:
    if word not in word_dict:
        word_dict[word] = 1
    else:
        word_dict[word] += 1
print(word_dict)

for key, value in word_dict.items():
    if value >= 3:
        print(f"{key}: {value}")
```
