```python
from pathlib import Path
import pandas as pd

BASE_DIR = Path.cwd()

file_path = BASE_DIR  / "store_sales.csv"

df = pd.read_csv(file_path)
```

```python
print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.dtypes)
```

```python
print(df.isna().sum())
```

```python
df["sales_amount"] = df["price"] * df["quantity"]

print(
    df[["product", "price", "quantity", "sales_amount"]]
)

print(df["sales_amount"].min())
print(df["sales_amount"].max())
print(df["sales_amount"].sum())

```

```python

print(df["category"].value_counts())
```

```python
print(df["channel"].value_counts())
```

```python
category_sales = df.groupby("category")["sales_amount"].sum()
print(category_sales)
```

```python

category_sales = (
    df.groupby("category")["sales_amount"]
      .sum()
      .sort_values(ascending=False)
)
print(category_sales)
```

```python

channel_sales = df.groupby("channel")["sales_amount"].sum()
print(channel_sales)
```

```python
# 온라인 판매만 놓고 총 매출은 얼마인가
online = df[df["channel"] == "온라인"]
print(online["sales_amount"].sum())
```

```python
# 판매 금액이 100,000원 이상인 기록은 무엇인가?
high_sales = df[df["sales_amount"] >= 100000]
print(high_sales[["product", "sales_amount"]])
```

```python
# 오프라인 판매가 선택했을 때 카테고리별 총 매출은 얼마인가?
df["sales_amount"] = df["price"] * df["quantity"]
df[df["channel"] == "오프라인"].groupby("category")["sales_amount"].sum()
```

```python
print(df["city"].isna().sum())
```

```python

category_summary = (
    df.groupby("category", as_index=False)["sales_amount"]
      .sum()
      .sort_values("sales_amount", ascending=False)
)
```

```python
category_summary.to_csv(
    "category_sales_summary.csv",
    index=False,
    encoding="utf-8-sig",
)
```
