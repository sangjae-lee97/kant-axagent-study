```python

class Student:
    def __init__(self, name, score, age):
        self.name = name
        self.score = score
        self.age = age

    def show_result(self):
        print(f"{self.name}: {self.score}점")

    def is_passed(self):
        return self.score >= 60
```

```python
student1 = Student("홍길동", 45, 30)
student2 = Student("강감찬", 100, 25)
student3 = Student("이감찬", 20, 15)

students = [student1, student2]
students.append(student3)

for student in students:
    print(f"{student.name}님의 점수는 {student.score}점 입니다.")

    # result = ""
    # if student1.is_passed() == True:
    #     result = "당당히 합격"
    # else:
    #     result = "아쉽지만 탈락"

    print(f"{student.name}님은 이번 시험에서 {"당당히 합격" if student.is_passed() == True else "아쉽지만 탈락"}하셨습니다.")
```
