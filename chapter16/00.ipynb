```python
from pathlib import Path

Base_DIR = ""

# if 실행환경이 파이썬(.py) 또는 노트북

if "__file__" in globals():

# .py파일일 경우
    BASE_DIR = Path(__file__).resolve().parents[2]

else:
# 현재 노트북이 실행 중인 폴더 기준

    BASE_DIR = Path.cwd()

# 만약 현재 노트북 파일이 위치한 폴더 자체가 기준점과 가깝다면 부모 단계 수를 조정하세요.
# 예: BASE_DIR = Path.cwd().parents[1] 혹은 Path.cwd()

print(BASE_DIR)

file_path = BASE_DIR / "data" / "chapter16" / "memo.txt"
```

```python
try:
    age = int(input("나이를 입력하세요: "))
    print(f"입력한 나이: {age}")
except ValueError:
    print("나이는 숫자로 입력해 주세요.")
```

```python
try:

    number = int(input("숫자를 입력하세요: "))
    result = 100 / number
    print(result)

except ValueError:
    print("정수를 입력해 주세요.")

except ZeroDivisionError:
    print("0으로 나눌 수 없습니다.")
```

```python
try:
    number = int(input("정수: "))

except ValueError:
    print("정수가 아닙니다.")

else:
    print(f"입력 성공: {number}")

finally:
    print("입력 처리를 마쳤습니다.")
```
