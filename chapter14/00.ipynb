```python
# 두 수(1~100까지)를 입력 받아서 사칙연산 결과를 반환하는 calc라는 함수를 만드시오.
# 4개 결과값을 한번에 반환(Tuple
# 사용자가 입력한 값이 숫자가 아닐경우 오류메시지를 출력하고 프로그램 종료

def calc(num1, num2):
    addition = num1 + num2
    subtraction = num1 - num2
    multiplication = num1 * num2
    division = num1/num2 if num2 != 0 else
    return addition, subtraction, multiplication, division

num1 = int(input("1부터 100까지의 숫자 중 첫번째 숫자를 입력하세요: "))
num2 = int(input("1부터 100까지의 숫자 중 두번째 숫자를 입력하세요: "))
addition, subtraction ,multiplication, division = calc(num1, num2)

print(f"더하기: {addition}, 빼기: {subtraction}, 곱하기: {multiplication}, 나누기: {division}")
```

```python
# 두 수(1~9까지)를 입력 받아서 사칙연산 결과를 반환하는 calc라는 함수를 만드시오.
# 4개 결과값을 한번에 반환(Tuple
# 사용자가 입력한 값이 숫자가 아닐경우 오류메시지를 출력하고 프로그램 종료

def calc(num1, num2):
    sum = num1 + num2
    minus = num1 - num2
    multi = num1 * num2
    div = num1 / num2 
    return sum, minus, multi, div

first = input("첫번째 숫자: ")
second = input("두번째 숫자: ")

lenth = len(first) + len(second)

if (lenth == 2 and ord(first) >= 49 and ord(first) <= 57) and (lenth == 2 and ord(second) >= 49 and ord(second) <= 57):
    # 제대로 둘다 숫자로 변환 가능
    result = calc(int(first),int(second))
    print(type(result))
    print(result)
else:
    # 잘못된 값이 입력된 경우
    print("입력한 값(숫자)를 확인하세요!")


```
